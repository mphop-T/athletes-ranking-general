# updated using hyperparameter tuning 

```python
# --- Reuse the exact grid you already defined ---
param_grid_large = {
    'num_leaves': np.arange(20, 200, 20),
    'max_depth': np.arange(4, 10),
    'learning_rate': np.linspace(0.01, 0.1, 10),
    'n_estimators': np.arange(500, 2000, 200),
    'min_child_samples': np.arange(10, 80, 10),
    'subsample': np.linspace(0.6, 1.0, 5),
    'colsample_bytree': np.linspace(0.6, 1.0, 5),
    'reg_alpha': np.linspace(0, 1.5, 6),
    'reg_lambda': np.linspace(0, 1.5, 6),
}


def validation_ndcg_at_10(model, X_val, y_val, val_group_sizes, event_col="Event_ID"):
    """
    Mean NDCG@5 on the validation partition, computed per event,
    matching the validation-NDCG@5 selection criterion already
    used for ListNet/ListMLE.
    """
    preds = model.predict(X_val)
    df = X_val.copy()
    df["_pred"] = preds

    if hasattr(y_val, "values"):
        df["_true"] = y_val.values
    else:
        df["_true"] = y_val

    scores = []
    for _, group in df.groupby(event_col):
        if len(group) < 2:
            continue
        y_true = group["_true"].to_numpy().reshape(1, -1)
        y_score = group["_pred"].to_numpy().reshape(1, -1)
        scores.append(ndcg_score(y_true, y_score, k=10))
    return np.mean(scores)


def clean_param_types(params):
    clean = {}
    for key, value in params.items():
        if isinstance(value, np.integer):
            clean[key] = int(value)
        else:
            clean[key] = float(value)
    return clean


def get_val_ndcg_10(result):
    return result["val_ndcg_10"]


def tune_lambdarank(X_train, y_train, train_group, X_val, y_val, val_group, n_iter=100, seed=42):
    sampler = ParameterSampler(param_grid_large, n_iter=n_iter, random_state=seed)
    results = []

    for params in sampler:
        clean_params = clean_param_types(params)

        model = lgb.LGBMRanker(
            objective="lambdarank",
            metric="map",
            boosting_type="gbdt",
            random_state=seed,
            **clean_params,
        )
        model.fit(
            X_train, y_train,
            group=train_group,
            categorical_feature=["Competitor", "Nationality", "Event_ID"],
        )

        score = validation_ndcg_at_10(model, X_val, y_val, val_group)
        results.append({"params": clean_params, "val_ndcg_10": score})
        print("val NDCG@10 =", round(score, 4), " | params =", clean_params)

    results.sort(key=get_val_ndcg_10, reverse=True)
    return results
```

# running updated lambdaRank hyperparameter tuning 

```python
# Run the search (men) 
search_results_men = tune_lambdarank(
    X_train_merged_df_men, y_train_merged_df_men, training_data_count_men,
    X_train_merged_df_men_validation, y_train_merged_df_men_validation, validation_data_count_men,
    n_iter=25, seed=42,
)
best_params_men = search_results_men[0]["params"]
print("Best params (men):", best_params_men)
print("Best validation NDCG@10 (men):", search_results_men[0]["val_ndcg_10"])
```

```python
# Run the search (women)
search_results_women = tune_lambdarank(
    X_train_merged_df_women, y_train_merged_df_women, training_data_count_women,
    X_train_merged_df_women_validation, y_train_merged_df_women_validation, validation_data_count_women,
    n_iter=25, seed=42,
)
best_params_women = search_results_women[0]["params"]
print("Best params (women):", best_params_women)
print("Best validation NDCG@10 (women):", search_results_women[0]["val_ndcg_10"])
```
