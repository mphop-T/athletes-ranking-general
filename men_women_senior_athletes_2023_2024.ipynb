```python
# calculate the cumulative best and shift by 1 down, ensuring that it 
# calculates using previous races only
# PB is Personal Best
merged_df_women["PB_before"] = (
    merged_df_women.groupby("Competitor")["Mark"]
             .apply(lambda x: x.cummin().shift())
             .reset_index(level=0, drop=True)
)
merged_df_women
```

```python
# calculate the cumulative best and shift by 1 down, ensuring that it 
# calculates using previous races only
# PB is Personal Best
merged_df_men["PB_before"] = (
    merged_df_men.groupby("Competitor")["Mark"]
             .apply(lambda x: x.cummin().shift())
             .reset_index(level=0, drop=True)
)
merged_df_men
```

```python
for label, df in [("Women", merged_df_women), ("Men", merged_df_men)]:
    total = len(df)
    no_prior_pb = df["PB_before"].isna().sum()
    pct_pb = no_prior_pb / total * 100
    print(f"{label}: {no_prior_pb} of {total} records with no prior competition history ({pct_pb:.2f}%)")
```

```python
# calculate the cumulative best and shift by 1 down, ensuring that it 
# calculates using previous races only
# SB is Seasonal Best (season is year in this case)
merged_df_women["SB_before"] = (
    merged_df_women.groupby(["Competitor", "Year"])["Mark"]
    .transform(lambda x: x.cummin().shift())
)
merged_df_women
```

```python
def convert_to_libsvm(df, file_path, feature_cols=None):
    # Determine feature columns (everything except Event_ID and Pos)
    if feature_cols is None:
        feature_cols = [c for c in df.columns if c not in ("Event_ID", "Relevance")]

    with open(file_path, "w") as f:
        # Group by Event_ID because it becomes qid
        for event_id, group in df.groupby("Event_ID"):

            for _, row in group.iterrows():

                relevance = int(row["Relevance"])          # relevance = Pos
                qid = int(row["Event_ID"])           # qid = Event_ID

                # Create features
                features = []
                for idx, col in enumerate(feature_cols, start=1):
                    val = row[col]
                    if pd.isna(val):
                        val = 0.0
                    features.append(f"{idx}:{float(val)}")

                # Write line
                f.write(f"{relevance} qid:{qid} " + " ".join(features) + "\n")

DESIRED_ORDER = [
    'Competitor', 'Nationality', 'Description', 'Event_Segment', 'Wind', 'Cat', 'Competition_Name', 'Competition_Group', 'Country', 'Key',
    'Age', 'Venue_Familiarity', 'Latitude', 'Longitude', 'Altitude', 'Month', 'Year', 'Temperature', 'Humidity',
    'ApparentTemperature', 'PressureMSL', 'SurfacePressure', 'Total_Participants', 'PB_before', 'SB_before', 'Recent_Avg_5', 'Recent_Best_5', 'Days_Since_Last', 'Weeks_Since_Last', 
    'Max_Rank']

train_df = X_train_merged_df_women_allRank.copy()
vali_df = X_train_merged_df_women_allRank_validation.copy()
test_df = X_test_merged_df_women.copy()

# compute feature_cols from train and pass it to both calls
feature_cols_from_train = [c for c in DESIRED_ORDER if c in train_df.columns and c not in ('Event_ID', 'Relevance')]
convert_to_libsvm(train_df, 'train_women.txt', feature_cols=feature_cols_from_train)
convert_to_libsvm(vali_df, 'vali_women.txt', feature_cols=feature_cols_from_train)
convert_to_libsvm(test_df,  'test_women.txt',  feature_cols=feature_cols_from_train)

train_df = X_train_merged_df_men_allRank.copy()
vali_df = X_train_merged_df_men_allRank_validation.copy()
test_df = X_test_merged_df_men.copy()

convert_to_libsvm(train_df, 'train_men.txt', feature_cols=feature_cols_from_train)
convert_to_libsvm(vali_df, 'vali_men.txt', feature_cols=feature_cols_from_train)
convert_to_libsvm(test_df,  'test_men.txt',  feature_cols=feature_cols_from_train)
```
