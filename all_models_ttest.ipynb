```python
import scipy.stats as stats
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import math
import pandas as pd 
```

```python
from sklearn.metrics import ndcg_score
from statsmodels.stats.multitest import multipletests
```

```python
X_test_merged_df_men_allRank = pd.read_excel('X_test_merged_df_men_allRank.xlsx')
X_test_merged_df_women_allRank = pd.read_excel('X_test_merged_df_women_allRank.xlsx')
X_test_merged_df_men_allRank=X_test_merged_df_men_allRank.drop(['Unnamed: 0'], axis=1)
X_test_merged_df_women_allRank=X_test_merged_df_women_allRank.drop(['Unnamed: 0'], axis=1)
```

```python
X_test_merged_df_men_copy_LambdaRank=pd.read_excel("X_test_merged_df_men_copy_LambdaRank.xlsx")
X_test_merged_df_women_copy_LambdaRank=pd.read_excel("X_test_merged_df_women_copy_LambdaRank.xlsx")
X_test_merged_df_men_copy_LambdaRank=X_test_merged_df_men_copy_LambdaRank.drop(['Unnamed: 0'], axis=1)
X_test_merged_df_women_copy_LambdaRank=X_test_merged_df_women_copy_LambdaRank.drop(['Unnamed: 0'], axis=1)
```

```python
X_test_merged_df_men_copy_Random_FR_results=pd.read_excel("X_test_merged_df_men_copy_Random_FR_results.xlsx")
X_test_merged_df_women_copy_Random_FR_results=pd.read_excel("X_test_merged_df_women_copy_Random_FR_results.xlsx")
X_test_merged_df_men_copy_Random_FR_results=X_test_merged_df_men_copy_Random_FR_results.drop(['Unnamed: 0'], axis=1)
X_test_merged_df_women_copy_Random_FR_results=X_test_merged_df_women_copy_Random_FR_results.drop(['Unnamed: 0'], axis=1)
```

```python
# ---- Full-list NDCG per event, per model, per gender ----

ndcg_full_listmle_men = []
for _, group in X_test_merged_df_men_allRank.groupby("Event_ID"):
    y_true = group["Relevance"].values
    y_score = group["listmle_men_pred"].values
    ndcg_full_listmle_men.append(ndcg_score([y_true], [y_score], k=len(group)))

ndcg_full_listnet_men = []
for _, group in X_test_merged_df_men_allRank.groupby("Event_ID"):
    y_true = group["Relevance"].values
    y_score = group["listnet_men_pred"].values
    ndcg_full_listnet_men.append(ndcg_score([y_true], [y_score], k=len(group)))

ndcg_full_lambdarank_men = []
for _, group in X_test_merged_df_men_copy_LambdaRank.groupby("Event_ID"):
    y_true = group["Relevance"].values
    y_score = group["LambdaRank_pred_tuned"].values
    ndcg_full_lambdarank_men.append(ndcg_score([y_true], [y_score], k=len(group)))

ndcg_full_randomforest_men = []
for _, group in X_test_merged_df_men_copy_Random_FR_results.groupby("Event_ID"):
    y_true = group["Relevance"].values
    y_score = group["Predicted_tuned"].values
    ndcg_full_randomforest_men.append(ndcg_score([y_true], [y_score], k=len(group)))

ndcg_full_listmle_women = []
for _, group in X_test_merged_df_women_allRank.groupby("Event_ID"):
    y_true = group["Relevance"].values
    y_score = group["listmle_women_pred"].values
    ndcg_full_listmle_women.append(ndcg_score([y_true], [y_score], k=len(group)))

ndcg_full_listnet_women = []
for _, group in X_test_merged_df_women_allRank.groupby("Event_ID"):
    y_true = group["Relevance"].values
    y_score = group["listnet_women_pred"].values
    ndcg_full_listnet_women.append(ndcg_score([y_true], [y_score], k=len(group)))

ndcg_full_lambdarank_women = []
for _, group in X_test_merged_df_women_copy_LambdaRank.groupby("Event_ID"):
    y_true = group["Relevance"].values
    y_score = group["LambdaRank_pred_tuned"].values
    ndcg_full_lambdarank_women.append(ndcg_score([y_true], [y_score], k=len(group)))

ndcg_full_randomforest_women = []
for _, group in X_test_merged_df_women_copy_Random_FR_results.groupby("Event_ID"):
    y_true = group["Relevance"].values
    y_score = group["Predicted_tuned"].values
    ndcg_full_randomforest_women.append(ndcg_score([y_true], [y_score], k=len(group)))


# Paired t-tests and Wilcoxon signed-rank tests, all six pairs, both genders 

t_listmle_listnet_men, p_listmle_listnet_men = stats.ttest_rel(ndcg_full_listmle_men, ndcg_full_listnet_men)

t_listmle_lambdarank_men, p_listmle_lambdarank_men = stats.ttest_rel(ndcg_full_listmle_men, ndcg_full_lambdarank_men)

t_listmle_randomforest_men, p_listmle_randomforest_men = stats.ttest_rel(ndcg_full_listmle_men, ndcg_full_randomforest_men)

t_listnet_lambdarank_men, p_listnet_lambdarank_men = stats.ttest_rel(ndcg_full_listnet_men, ndcg_full_lambdarank_men)

t_listnet_randomforest_men, p_listnet_randomforest_men = stats.ttest_rel(ndcg_full_listnet_men, ndcg_full_randomforest_men)

t_lambdarank_randomforest_men, p_lambdarank_randomforest_men = stats.ttest_rel(ndcg_full_lambdarank_men, ndcg_full_randomforest_men)

t_listmle_listnet_women, p_listmle_listnet_women = stats.ttest_rel(ndcg_full_listmle_women, ndcg_full_listnet_women)

t_listmle_lambdarank_women, p_listmle_lambdarank_women = stats.ttest_rel(ndcg_full_listmle_women, ndcg_full_lambdarank_women)

t_listmle_randomforest_women, p_listmle_randomforest_women = stats.ttest_rel(ndcg_full_listmle_women, ndcg_full_randomforest_women)

t_listnet_lambdarank_women, p_listnet_lambdarank_women = stats.ttest_rel(ndcg_full_listnet_women, ndcg_full_lambdarank_women)

t_listnet_randomforest_women, p_listnet_randomforest_women = stats.ttest_rel(ndcg_full_listnet_women, ndcg_full_randomforest_women)

t_lambdarank_randomforest_women, p_lambdarank_randomforest_women = stats.ttest_rel(ndcg_full_lambdarank_women, ndcg_full_randomforest_women)
```

```python
print("=== Men ===")
print(f"ListMLE vs ListNet:        t = {t_listmle_listnet_men:.2f}, p = {p_listmle_listnet_men:.2f}")
print(f"ListMLE vs LambdaRank:     t = {t_listmle_lambdarank_men:.2f}, p = {p_listmle_lambdarank_men:.2f}")
print(f"ListMLE vs Random Forest:  t = {t_listmle_randomforest_men:.2f}, p = {p_listmle_randomforest_men:.2f}")
print(f"ListNet vs LambdaRank:     t = {t_listnet_lambdarank_men:.2f}, p = {p_listnet_lambdarank_men:.2f}")
print(f"ListNet vs Random Forest:  t = {t_listnet_randomforest_men:.2f}, p = {p_listnet_randomforest_men:.2f}")
print(f"LambdaRank vs Random Forest: t = {t_lambdarank_randomforest_men:.2f}, p = {p_lambdarank_randomforest_men:.2f}")

print("\n=== Women ===")
print(f"ListMLE vs ListNet:        t = {t_listmle_listnet_women:.2f}, p = {p_listmle_listnet_women:.2f}")
print(f"ListMLE vs LambdaRank:     t = {t_listmle_lambdarank_women:.2f}, p = {p_listmle_lambdarank_women:.2f}")
print(f"ListMLE vs Random Forest:  t = {t_listmle_randomforest_women:.2f}, p = {p_listmle_randomforest_women:.2f}")
print(f"ListNet vs LambdaRank:     t = {t_listnet_lambdarank_women:.2f}, p = {p_listnet_lambdarank_women:.2f}")
print(f"ListNet vs Random Forest:  t = {t_listnet_randomforest_women:.2f}, p = {p_listnet_randomforest_women:.2f}")
print(f"LambdaRank vs Random Forest: t = {t_lambdarank_randomforest_women:.2f}, p = {p_lambdarank_randomforest_women:.2f}")
```
